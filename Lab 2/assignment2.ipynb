```python
import pandas as pd
df = pd.read_excel('Heart (1).xlsx')
duplicates = df.duplicated().sum()
print("Duplicate records:", duplicates)
df = df.drop_duplicates()
print("After removing duplicates:", df.duplicated().sum())
print(df.head())
print(df.tail())
```

```python
import pandas as pd
df = pd.read_excel('Heart (1).xlsx')
Q1 = df['Chol'].quantile(0.25)
Q3 = df['Chol'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['Chol'] < lower_bound) | (df['Chol'] > upper_bound)]
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", outliers.shape[0])
print(outliers[['Chol']])
```

```python
import pandas as pd
df = pd.read_excel('Heart (1).xlsx')
Q1 = df['RestBP'].quantile(0.25)
Q3 = df['RestBP'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
median_value = df['RestBP'].median()
df['RestBP'] = df['RestBP'].apply(lambda x: median_value if x < lower_bound or x > upper_bound else x)
print(df['RestBP'].describe())
```

```python
import pandas as pd
df = pd.read_excel('Heart (1).xlsx')
categorical_cols = df.select_dtypes(include='object').columns
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
print(df_encoded.head())
print(df_encoded.tail())
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
df = pd.read_excel('Heart (1).xlsx')
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
print(df.head())
print(df.tail())
```
